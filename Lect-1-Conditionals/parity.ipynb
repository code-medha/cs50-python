# Parity Checker Even or Odd

This notebook walks through different ways to write a small Python program that checks whether a number is **even** or **odd**.

## What is parity?

- **Even** — divisible by 2 (e.g. 0, 2, 4, 10)
- **Odd** — not divisible by 2 (e.g. 1, 3, 7, 9)

We use the **modulo operator** (`%`) to test this:

```python
x % 2 == 0   # True if x is even, False if x is odd
```

| Expression | Meaning |
|---|---|
| `4 % 2` | Remainder when 4 is divided by 2 → `0` |
| `5 % 2` | Remainder when 5 is divided by 2 → `1` |

If the remainder is `0`, the number is even.



> **Note:** You do **not** need a `main()` function in Python. It is optional and mainly helps organize larger programs. For small scripts and notebooks, calling functions directly is fine.

## Step 1: Read a number from the user

We start by asking for input and converting it to an integer.

- `input()` — reads text from the user (always returns a string)
- `int()` — converts that string to a whole number

The result is stored in `x`, which we use in later cells.

**Run this cell first** — later cells depend on the variable `x`.

````python
x = int(input("What's x? "))
````

## Step 2: Organize with `main()` and a helper function

Here we split the program into two responsibilities:

| Function | Role |
|---|---|
| `main()` | Runs the program — gets input and prints the result |
| `is_even(n)` | Checks parity — returns `True` if even, `False` if odd |

### Why use a helper function?

- **Reusable** — call `is_even(4)` from anywhere without rewriting the logic
- **Testable** — you can check `is_even(4)` without running the whole program
- **Readable** — `if is_even(x):` reads like plain English

### Explicit return with `if` / `else`

This version returns `True` or `False` using an explicit conditional:

```python
return True if n % 2 == 0 else False
```

That works, but the next section shows a shorter equivalent.

````python
def main():
    x = int(input("What's x"))

    if is_even(x):
        print("even")
    else:
        print("odd")

def is_even(n):
    return True if n % 2 == 0 else False         

main()           

````

## Step 3: Simplify `is_even()` — return the condition directly

The expression `n % 2 == 0` already evaluates to `True` or `False`. There is no need for an extra `if` / `else` inside the function.

```python
# Verbose (works, but unnecessary)
return True if n % 2 == 0 else False

# Cleaner — same behavior
return n % 2 == 0
```

### `return` vs `print`

| | `print()` | `return` |
|---|---|---|
| **Purpose** | Shows output on screen | Sends a value back to the caller |
| **Use when** | You want to display something | You want the function to *answer* a question |

Here, `is_even()` **returns** a boolean. `main()` decides what to **print** based on that value.

This separation is a common pattern in real Python code.

````python
def main():
    x = int(input("What's x"))

    if is_even(x):
        print("even")
    else:
        print("odd")

def is_even(n):
    return n % 2 == 0        

main()           
````

## Step 4: Without `main()` — call the function directly

You can skip `main()` entirely and call `is_even()` on its own.

When you run this cell, Python evaluates `is_even(x)` and displays the return value (`True` or `False`) as the cell output.

This is valid Python — `main()` is optional, especially in notebooks and small scripts.

````python
def is_even(x):
    return x % 2 == 0

is_even(x)        
````

## Step 5: Print "Even" or "Odd" using the return value

The function returns a boolean; the code *outside* the function handles printing.

```python
if is_even(x):
    print("Even")
else:
    print("Odd")
```

This is the recommended split:

1. **`is_even(x)`** — answers: "Is this number even?" → `True` / `False`
2. **`if` / `else` block** — decides what message to show the user

---

## Summary

| Approach | When to use |
|---|---|
| Top-level code (no functions) | Very short, one-off scripts |
| Function that **prints** inside | Quick demos; harder to reuse |
| Function that **returns** + print outside | Best for reusable, testable logic |
| `main()` + `if __name__ == "__main__":` | Standard for `.py` script files |

For CS50 exercises and notebooks, any of the function-based approaches above is fine. As programs grow, prefer **returning values** from helpers and keeping **printing** in the code that calls them.

````python
def is_even(x):
    return x % 2 == 0

if is_even(x):
    print("Even")
else:
    print("Odd")    
````
