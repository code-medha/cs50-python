Learn about nested loops

Print column

```python
def main():
    print_column(3)


def print_column(height):
    for _ in range(height):
        print("#")

main()          
```

print row
By default, print("?") is the same as print("?", end="\n"), so each "?" lands on its own line. Here, end="" means `print ? and then nothing,` so the next print continues on the same line.

```python
from os import sep


def main():
    print_row(4)


def print_row(width):
    for _ in range(width):
        print("?", end="") 

main()           
```

create 3 * 3 brick

```python
def print_square(size):

    for i in range(size):
        for j in range(size):
            print("#", end="")

        print()    

print_square(3)            
```

How does nested loops work?
The outer loop does one step, then the inner loop runs all the way through, then the outer loop takes the next step. That is why you see i stay at 0 while j goes 0, 1, 2.

What “nested” means

Python does not run both loops at the same time. It treats the inner for as the body of the outer for. For each value of i, it must finish that whole body before i changes.


always remeber that i and j are counters. Python does not print i, j, or “0 0”. Those numbers only control how many times the body runs.

another simple way to do it

```python
def print_square(size):

    for i in range(size):
        print("#" * size)

print_square(3)        

```
