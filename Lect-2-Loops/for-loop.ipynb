Before we get started with for loop, we should understand 'list' data structure.
So what is a list?
A list is a data structure that allows you to store multiple values under one variable/reference.


marks = [85, 72, 91]

don't think: "marks contains three integers."
instead: "marks refers to a collection that contains references to three Python objects."

```python
for i in [0, 1, 2]:
    print("Deep Seek")
```

The above program is poorly designed. Instead of declaring the list, we can use Python's 'range' 

```python
for i in range(5):
    print("Open Code")
```

There's a pythonic way to improve the above function. i.e you don't use the variable anywhere in the code, then you can use underscore "_"

```python
for _ in range(7):
    print("Kimi")
```

Now let's modify the above program to ask the user to repeat the word based n the user input

```python
def main():
    while True:
        n = int(input("What's the number "))
        if n < 0:
            continue
        else:
            break

    do_repeat(n)   
        
def do_repeat(x):
    for _ in range(x):
        print("meow")
    


main()

```

Improvisation

```python
def main():
    while True:
        n = int(input("What's the number "))
        if n > 0:
            do_repeat(n)           
            return n
  
        
def do_repeat(x):
    for _ in range(x):
        print("meow")


main()
```

understanding the flow:
n = 5, then do_repeat(5). It is the value 5 that gets passed to do_repeat, not the variable n itself. Then Python enters do_repeat. We have defined def do_repeat(x): where x is a parameter. When you called do_repeat(5), then Python puts that value into x. So inside do_repeat, x → 3

return can both exit a function and send a value back.
