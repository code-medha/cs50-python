Iterating over lists

```python
students = ["Hermonie", "Harry", "Ron"]

for student in students:
    print(student)
```

suppose you want to iterate using numbers like 0, 1, 2 etc.
so you can use len fucntion, which will tell the length of the list

Always remember range only take intergers, it doesn't take strings. In such cases you can make use of len function.

len(students) is 3. So range(3)

```python
students = ["Hermonie", "Harry", "Ron"]

for i in range(len(students)):
    print(students[i])
```
