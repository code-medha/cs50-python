Iterating over lists

```python
students = ["Hermonie", "Harry", "Ron"]

for student in students:
    print(student)
```

suppose you want to iterate using numbers like 0, 1, 2 etc.
so you can use len fucntion, which will tell the length of the list

Always remember range only take intergers, it doesn't take strings. In such cases you can make use of len function.

len(students) is 3. So range(3)

```python
students = ["Hermonie", "Harry", "Ron"]

for i in range(len(students)):
    print(students[i])
```

Looping Dictinaries

```python
children = {
    "Darshan": "Green",
    "David": "Blue",
    "Rita": "Green",
    "Abhi": "Yellow"
}

print(children)
print(children["Darshan"])
```

print(child) → prints the key (the name)
print(children[child]) → uses that key to look up and print the value (the color)

When you write for child in children, Python gives you only the keys, not the values.
To get the value, you have to look it up with children[child] — same as when you wrote children["Darshan"] earlier and got "Green".



```python
children = {
    "Darshan": "Green",
    "David": "Blue",
    "Rita": "Green",
    "Abhi": "Yellow"
}

for child in children:
    print(child)
    print(children[child])
```

Alternatively, you can use .items() method. 

```python
children = {
    "Darshan": "Green",
    "David": "Blue",
    "Rita": "Green",
    "Abhi": "Yellow"
}

for name, color in children.items():
    print(f'{name} belongs to {color} house')
```

Dictinary in a list

```python
students = [
    {"name": "Hermonie", "house": "Gryffindor", "patronus": "Otter"},
    {"name": "Harry", "house": "Gryffindor", "patronus": "Stag"},
    {"name": "Ron", "house": "Slytherin", "patronus": None }
]

for student in students:
    print(student["name"], student["house"], sep=", ")
```

Always remember Looping a dict yields keys, wheres looping a list yields each element. Lists are indexed only by integers (or slices). So Python raises TypeError: list indices must be integers or slices, not dict.

If you want index-based access on the list:

```python
students = [
    {"name": "Hermonie",
     "house": "Gryffindor", 
     "patronus": "Otter"},

    {"name": "Harry", 
    "house": "Gryffindor", 
    "patronus": "Stag"},

    {"name": "Ron", 
    "house": "Slytherin",
     "patronus": None }
]


for i in range(len(students)):
    print(students[i]["name"])
```
