'While' let's us do again and again. A while loop only runs its body when the condition is True.

```python
x = 3

while x != 0:
    print("machine learning")
    x = x - 1
```

If we want to start from ascending order and not using 'not equal' to.

```python
x = 1

while x <= 3:
    print("Deep Learning")
    x= x + 1
```

As per the python convention, we need to start counting from 0. So if we assign x=0, then how does the logic will change?

```python
x = 0

while x < 3:
    print("RAG")
    x = x + 1
```

short-hand notation of writing x = x+1



```python
x = 0

while x < 3:
    print("Agent")
    x += 1
```
