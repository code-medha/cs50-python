Let's learn try and except syntax.
Except meaning: (Preposition or Conjunction) Use except when you want to show an exclusion or an exception to a rule.

````python
try:
    x = int(input("What's x? "))
    print(f"x is {x}")
except ValueError:
    print(" x is not an integer")
````

code improvisation
In Python's try-except-else block, the else code only runs if the try block succeeds without raising any exceptions.

If valid input is given, x gets successfully assigned, and the else block executes print(f"x is {x}").

If invalid input is given, the except block handles the error, and the else block is skipped completely, preventing Python from attempting to access the undefined x.

````python
try:
    x = int(input("What's x? "))
except ValueError:
    print("x is not an integer")
else:
    print(f"x is {x}")            
````

I always get confused with paramter and agrument when defining a function.
**Easy Rule of Thumb:**
Parameter = Placeholder (used when defining the function)
Argument = Actual value (used when calling the function)

A return statement has only one job: it exits/terminates the entire function and sends a value back to the caller. 
Because it exits the function, any loop running inside that function is automatically terminated as a side effect. It doesn't just "break" the loop; it ends the entire function execution immediately.

````python
def main():
    x = get_int()
    print(f"x is {x}")

def get_int():

    while True:
        try:
            x = int(input("What's x? "))
        except ValueError:
            print("x is not an integer")
        else:
            return x

main() 
````

````python
def main():
    x = get_int("what's x ")
    print(f"x is {x}")

def get_int(prompt):

    while True:
        try:
            return int(input(prompt))
        except ValueError:
            pass

main() 
````

Always remember:
```
1. DEFINING (Create a placeholder box):
   def get_int( prompt ):
                 ▲
                 └─ Parameter (The box is empty)

2. CALLING (Pass a value into that box):
   x = get_int( "what's x " )
                    ▲
                    └─ Argument (The value we put in the box)
```
