importing libraries

when you use `import` you're importing everything that is available in that function.

```python
import random

coin = random.choice(["heads", "tails"])

print(coin)
```

`from` let's you import the specific function from the module

```python
from random import choice

coin = choice(["heads", "tails"])

print(coin)
```

```python
import random


cards = ["queen", "king", "joker"]

random.shuffle(cards)
print(cards)

for card in cards:
    print(card)
```
