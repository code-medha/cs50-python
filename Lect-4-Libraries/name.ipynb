Handle errors 

```python
import sys

try:
    print("hello, my name is", sys.argv[1])
except IndexError:
    print("Too few arguments")    
```

Can be more defensive in writing the code. Check for multiple names, no names

```python
import sys


if len(sys.argv) > 2:
    print("Too many arguments")
elif len(sys.argv) < 2:
    print("Too less arguments") 
else:
    print("Hello, my name is", sys.argv[1])      
```

len(sys.argv) gets the number of command-line arguments.
len(sys.argv[1]) gets the length of the string passed as the first argument (e.g., "Alice" has a length of 5).

sys.argv[0] is alwyas the file name.
sys.argv[0]   # hello.py
sys.argv[1]   # Darshan

sys.exit

```python
import sys

if len(sys.argv) > 2:
    sys.exit("Too many arguments")
elif len(sys.argv) < 2:
    sys.exit("Too less arguments")

print("Hello, my name is", sys.argv[1])        
```

multiple arguments

```python
import sys

if len(sys.argv) > 2:
    sys.exit("Too many arguments")

for arg in sys.argv:
    print("Hello, my name is", arg)
```

the above program will induce a bug, ie it will loop through the sys.argv[0] as well. so we have to make use of slices.

```python
import sys

if len(sys.argv) > 2:
    sys.exit("Too many arguments")

for arg in sys.argv[1:]:
    print("Hello, my name is", arg)
```
